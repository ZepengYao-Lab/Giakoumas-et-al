# Experiment: Full Hops Figure


This notebook replaces `Full_hops_figure.ipynb` with a shared Sankey builder. It recreates the portrait multi-panel hops figure across the PhN, PSO-SA, and MxLbN datasets.


```python
from pathlib import Path
import sys


def find_project_root(start: Path) -> Path:
    for candidate in (start, *start.parents):
        if (candidate / "src").is_dir() and (candidate / "data").is_dir():
            return candidate
    raise RuntimeError("Could not locate the standalone project root.")


PROJECT_ROOT = find_project_root(Path.cwd().resolve())
SRC = PROJECT_ROOT / "src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from giakoumas_connectome.data import discover_project_root, load_repository_tables
from giakoumas_connectome.hops import (
    flatten_hop_collections,
    load_default_hop_collections,
    plot_stacked_sankey_panels,
    save_plotly_figure,
)
```

## Build The Sankey Panels


```python
PROJECT_ROOT = discover_project_root(Path.cwd())
OUTPUT_DIR = PROJECT_ROOT / 'output' / 'full_hops_figure'
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

tables = load_repository_tables(PROJECT_ROOT)
collections = load_default_hop_collections(PROJECT_ROOT)
panels = flatten_hop_collections(collections)

len(panels), [label for label, _ in panels[:5]]
```

## Full Portrait Sankey Figure


```python
figure = plot_stacked_sankey_panels(
    panels,
    tables.connections,
    tables.classification_other,
    title='All Sankey Panels in One Portrait Figure',
)
figure
```

## Exported Outputs


```python
figure_dir = OUTPUT_DIR / 'figures'
figure_dir.mkdir(parents=True, exist_ok=True)

svg_path = save_plotly_figure(
    figure,
    figure_dir / 'all_sankeys_portrait.svg',
    width=534,
    height=300 * len(panels),
)
html_path = figure_dir / 'all_sankeys_portrait.html'
figure.write_html(html_path)

[str(svg_path.relative_to(PROJECT_ROOT)), str(html_path.relative_to(PROJECT_ROOT))]
```
