# Experiment: Coconatfly PhN Cosine


This notebook is a streamlined R refactor of `2_Coconatfly_PhN_Cosine.ipynb`. It uses a shared Coconatfly heatmap workflow driven by a copied selected-population CSV.


```r
find_project_root <- function(start = getwd()) {
  current <- normalizePath(start, winslash = "/", mustWork = TRUE)
  repeat {
    if (dir.exists(file.path(current, "r")) && dir.exists(file.path(current, "data"))) {
      return(current)
    }
    parent <- dirname(current)
    if (identical(parent, current)) {
      stop("Could not locate the standalone project root.")
    }
    current <- parent
  }
}

PROJECT_ROOT <- find_project_root()
source(file.path(PROJECT_ROOT, "r", "bootstrap.R"))
gc_bootstrap()
options(repr.plot.res = 300)

PROJECT_ROOT
```

## Build The Shared Analysis


```r
analysis <- gc_run_coconatfly_analysis("phn", project_root = PROJECT_ROOT)
print(analysis$summary)
```

## Inline Heatmap


```r
gc_plot_heatmap(analysis)
```
