# Experiment: Coconatfly aPhN Cosine


This notebook is a streamlined R refactor of `3_Coconatfly_aPhN_Cosine.ipynb`. It rebuilds the three final analyses in that notebook: `aPhN`, `aPhN1`, and `aPhN2`.


```r
find_project_root <- function(start = getwd()) {
  current <- normalizePath(start, winslash = "/", mustWork = TRUE)
  repeat {
    if (dir.exists(file.path(current, "r")) && dir.exists(file.path(current, "data"))) {
      return(current)
    }
    parent <- dirname(current)
    if (identical(parent, current)) {
      stop("Could not locate the standalone project root.")
    }
    current <- parent
  }
}

PROJECT_ROOT <- find_project_root()
source(file.path(PROJECT_ROOT, "r", "bootstrap.R"))
gc_bootstrap()
options(repr.plot.res = 300)

PROJECT_ROOT
```

## aPhN


```r
aphn_result <- gc_run_coconatfly_analysis("aphn", project_root = PROJECT_ROOT)
print(aphn_result$summary)
gc_plot_heatmap(aphn_result)
```

## aPhN1


```r
aphn1_result <- gc_run_coconatfly_analysis("aphn1", project_root = PROJECT_ROOT)
print(aphn1_result$summary)
gc_plot_heatmap(aphn1_result)
```

## aPhN2


```r
aphn2_result <- gc_run_coconatfly_analysis("aphn2", project_root = PROJECT_ROOT)
print(aphn2_result$summary)
gc_plot_heatmap(aphn2_result)
```

## Exported Outputs


```r
paths <- list(
  aphn = unlist(gc_save_analysis_outputs(aphn_result)),
  aphn1 = unlist(gc_save_analysis_outputs(aphn1_result)),
  aphn2 = unlist(gc_save_analysis_outputs(aphn2_result))
)
print(paths)
```
