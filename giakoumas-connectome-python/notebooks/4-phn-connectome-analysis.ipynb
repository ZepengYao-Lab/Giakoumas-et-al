# Experiment: PhN Connectome Analysis


This notebook is intentionally thin. It uses the shared package code in `src/giakoumas_connectome/` to rebuild the copied PhN workflow and save standard tables plus figures under `output/`.


```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display


def find_project_root(start: Path) -> Path:
    for candidate in (start, *start.parents):
        if (candidate / "src").is_dir() and (candidate / "data").is_dir():
            return candidate
    raise RuntimeError("Could not locate the standalone project root.")


PROJECT_ROOT = find_project_root(Path.cwd().resolve())
SRC = PROJECT_ROOT / "src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

pd.options.display.max_columns = 100
pd.options.display.max_rows = 200

PROJECT_ROOT
```

```python
from giakoumas_connectome.config import get_workflow_config
from giakoumas_connectome.reports import build_workflow_report, export_standard_figures, save_report_tables
from giakoumas_connectome.plots import (
    plot_first_order_heatmap,
    plot_input_synapse_counts,
    plot_location_outputs,
    plot_nt_distribution,
    plot_order_counts,
    plot_second_order_interset_heatmap,
    plot_second_order_upset,
    plot_second_to_third_order_heatmap,
    plot_set_to_second_order_heatmap,
    plot_superclass_distribution,
    plot_third_order_upset,
    plot_top_regions,
)
```

## Build The Shared Report


```python
workflow = get_workflow_config("phn")
report = build_workflow_report(workflow, project_root=PROJECT_ROOT)
report = save_report_tables(report)
figure_paths = export_standard_figures(report)
print(f"Output directory: {report.output_dir}")
report.summary
```

## Summary Tables


```python
display(report.summary)
display(report.first_order_set_matrix)
```

## First-Order Overview


```python
fig, _ = plot_first_order_heatmap(
    report.first_order_set_matrix,
    title=f"{report.workflow.display_name}: first-order set-to-set synapses",
    display_labels=list(report.workflow.set_labels),
)
plt.show()

fig, _ = plot_input_synapse_counts(report)
plt.show()

fig, _ = plot_order_counts(report)
plt.show()
```

```python
fig, _ = plot_location_outputs(report, order="second")
plt.show()

fig, _ = plot_superclass_distribution(report, order="second")
plt.show()

fig, _ = plot_nt_distribution(report, order="second", location="local")
plt.show()

fig, _ = plot_nt_distribution(report, order="second", location="outside_SEZ")
plt.show()

fig, _ = plot_top_regions(report, order="second")
plt.show()
```
